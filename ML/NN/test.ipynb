```python
import numpy as np
import matplotlib.pyplot as plt
import numpy.random as r
from sklearn.datasets import load_digits
from neural_network import NN
from sklearn.metrics import accuracy_score
```

```python
# loading the dataset
data = load_digits()
X = data.data
y = data.target
img = data.images
n,m = X.shape
print(n,m)
```

```python
# splitting the data into training and testing set
indices = np.arange(n)
r.shuffle(indices)
N = 1000
X_train = X[indices[0:N]]
y_train = y[indices[0:N]]
img_train = img[indices[0:N]]
X_test = X[indices[N:]]
y_test = y[indices[N:]]
img_test = img[indices[N:]]
```

```python
# training the model
m = NN(max_iter=300,learning_rate=0.1,n_nodes=[100])
m.fit(X_train,y_train)
```

```python
y_pred, y_pred_prob = m.predict(X_test)
accuracy = accuracy_score(y_pred, y_test)
print(accuracy)
```
