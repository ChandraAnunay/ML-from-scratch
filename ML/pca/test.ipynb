```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_sample_images
```

```python
data = load_sample_images()
images = data.images
filenames = data.filenames
# china, flower = np.array(images[0]), np.array(images[1])
# choose an image
img = images[0]
```

```python
print(img.shape)
```

From above, we find that there are two images each of size 427 x 640 pixels

```python
print('Original Image')

plt.imshow(img)
plt.show()
```

```python
print('Cropped Image')

cropped_flower = img[30:400,125:500,:]
plt.imshow(cropped_flower)
plt.show()
```

```python
print('Colour clipped image')

clipped_flower = img.copy()
clipped_flower[clipped_flower <= 75] = 75
# try these
# clipped_flower[clipped_flower <= 100] = 50
# clipped_flower[clipped_flower <= 50] = 100
plt.imshow(clipped_flower)
plt.show()
```

```python
print('Red color removed')

no_red = img.copy()
no_red[:,:,0] = 0
plt.imshow(no_red)
plt.show()
```

```python
print('Green color removed')

no_green = img.copy()
no_green[:,:,1] = 0
plt.imshow(no_green)
plt.show()
```

```python
print('Blue color removed')

no_blue = img.copy()
no_blue[:,:,2] = 0
plt.imshow(no_blue)
plt.show()
```

```python
print('Only red color present')

only_red = img.copy()
only_red[:,:,[1,2]] = 0
plt.imshow(only_red)
plt.show()
```

```python
print('Only green color present')

only_green = img.copy()
only_green[:,:,[0,2]] = 0
plt.imshow(only_green)
plt.show()
```

```python
print('Only blue color present')

only_blue = img.copy()
only_blue[:,:,[0,1]] = 0
plt.imshow(only_blue)
plt.show()
```

```python
print('grayscale image')

gray = img.copy()
for i in range(len(gray)):
    for j in range(len(gray[0])):
        mean = np.mean(gray[i][j])
        for k in range(3):
            gray[i][j][k] = mean
plt.imshow(gray)
plt.show()
```
